# Working with the FormFields Class: Export Modifiable Fields to CSV

```python
import pymupdf
```

```python
doc = pymupdf.open("Customer-Info-1.pdf")
ff = doc.get_form_fields(repair=True, verbose=True)
```

Display terminal fields and their values.

```python
for name,entry in ff.fields.items():
    if entry["terminal"]:
        print(f"{name:<25} '{entry['value']}'")
```

Export fields that have a value ("terminal fields") to a CSV file.

```python
ff.to_csv("temp.csv")
```
