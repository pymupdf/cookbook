# Working with the FormFields Class: Import CSV File with Value Changes

```python
import pymupdf
```

```python
doc = pymupdf.open("Customer-Info-1.pdf")
ff = doc.get_form_fields(repair=True, verbose=True)
```

Show all terminal fields and their current values before the import.

```python
for name,entry in ff.fields.items():
    if entry["terminal"]:
        print(f"{name:<25} '{entry['value']}'")
```

Import the CSV file with any field value changes.

```python
ff.from_csv("temp.csv")
```

Show fields again to see the changes:

We have changed `Customer.Name.First`, `Customer.Name.Last` and `Customer.MaritalStatus`; the field `Customer.Fullname` has been updated automatically via JavaScript.
